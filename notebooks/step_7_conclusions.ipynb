# Step 7 - Connecting the Model to the Real-World Use Case

This section connects the technical results to the original business challenge: helping vineyard teams prioritise which grapevines require expert inspection, while keeping diagnosis and treatment decisions with qualified agronomists.

## Business challenge

Vineyard managers and agronomists need to monitor grapevines for visible signs of disease. Timely identification can support earlier assessment and treatment, but inspecting large vineyard areas requires time and specialist attention. The proposed role of the CNN is to analyse systematically collected leaf photographs and provide a triage signal that helps teams decide which vines should receive expert follow-up first.

The operational question is whether the model can identify likely disease symptoms reliably enough to support inspection prioritisation without creating an unacceptable number of missed cases or unnecessary reviews.

## Model performance

The CNN evaluated here is the dropout-0.40 configuration selected in Step 5 because it achieved the lowest validation loss among the tested configurations.

On the corrected held-out test set, the selected CNN achieved 82.04% accuracy, compared with a 45.30% majority-class baseline. Its weighted F1-score was 81.59%, while the lower macro F1-score of 68.05% showed that performance was uneven across the four classes. Healthy and Esca achieved F1-scores of 95.60% and 81.66%, whereas Black Rot and Leaf Blight achieved 50.00% and 44.94%. Recall was particularly limited for Black Rot (46.67%) and Leaf Blight (40.00%).

For the proposed triage decision, grouping the predictions into Healthy versus Disease produced 98.99% disease recall, 94.23% precision and a 96.55% F1-score. The model flagged 196 of 198 diseased images and missed two. It also flagged 12 healthy images unnecessarily. Under this decision rule, 57.46% of test images would be referred for expert review.

The error analysis showed that mistakes were concentrated in a small number of disease-class pairs, particularly Leaf Blight predicted as Esca. Seven incorrect predictions were made with confidence of at least 80%, demonstrating that model confidence cannot be treated as proof that a classification is correct.

## Proposed real-world workflow

1. Vineyard workers photograph grapevine leaves during routine monitoring, following a systematic sampling protocol. It is important to have a systematic approach to sampling to avoid the photographs depending uniquely on the workers decisions, since this would leave the sampling decision in the hands of non-experts.
2. The CNN processes each image and returns probabilities for Healthy, Black Rot, Esca and Leaf Blight. These predictions are used as a triage signal.
3. Vines associated with images flagged as potentially diseased are prioritised for in-person review.
4. An agronomist inspects the vine and independently decides whether to continue monitoring, investigate further or recommend treatment.
5. The expert-validated outcome is recorded so model performance, missed cases, false alarms and workload can be monitored over time. It can also be used to improve the model later on.

Routine vineyard monitoring would continue regardless of the model's prediction. The CNN would influence inspection priority and could never replace inspection or determine treatment.

## Limitations and controls

The current evidence is limited by class imbalance, uncertain dataset labels and geographical differences. Folder-derived classes were used as provisional targets because the supplied metadata conflicts systematically with some image folders. The cleaned data come from Indian vineyards, and performance has not been tested on images collected in Portuguese vineyards under the intended operating conditions.

The model can miss disease and can confuse specific diseases, sometimes with high confidence. Agronomist review must therefore remain mandatory, and probability scores should not be used as a diagnostic tool. Performance should be monitored by disease class as well as through the Healthy-versus-Disease triage measure. Any pilot should use expert-validated Portuguese images and predefined criteria for missed cases, false alarms and workload.

## Pilot recommendation

The results support a limited parallel pilot but are not good enough for operational deployment or autonomous diagnosis. 
- During the first 30 days, the team would collect expert-labelled Portuguese vineyard images. Alternatively, the original dataset from the article Grapevine Leaf Disease Dataset (https://pmc.ncbi.nlm.nih.gov/articles/PMC13098524/) could be procured by contacting the authors. Besides acquiring a reliable local dataset, the team would measure the time and disease-detection performance of the existing inspection process.
- During the following 30 days, the CNN would run alongside normal inspections without changing treatment decisions. The team would measure recall by disease, false alarms, missed cases, the proportion referred for review and agronomist workload.
- After 90 days, the pilot results would be compared with the manual-inspection baseline. The team could then decide whether to refine, extend or discontinue the pilot based on whether inspection prioritisation improves without compromising disease detection.

## Expected value and conclusion

The CNN has the potential to focus expert attention on vines that are more likely to show disease symptoms, supporting earlier follow-up and more targeted use of specialist time. However, the project would need to demonstrate time savings, cost reductions and reliable disease-specific diagnosis in Portuguese vineyards. These benefits would have to be measured during a controlled pilot rather than assumed from the test results.

Overall, the project demonstrates a promising decision-support proof of concept. Its strongest current use is Healthy-versus-Disease triage, while its weaker discrimination between specific diseases prevents a solid basis for diagnosis. The appropriate next decision would be to continue with a limited, expert-supervised pilot using data from the North of Portugal, while retaining human control over all diagnosis and treatment decisions.
