# Step 4 – Why Deep Learning is Appropriate

## Objective

This notebook explains why Deep Learning, and specifically a Convolutional Neural Network (CNN), is an appropriate approach for the grapevine disease image-classification problem defined in this project.

## 1. Image data contains spatial information

The input data consists of RGB images of grapevine leaves. In image-classification problems, neighbouring pixels are related and their spatial arrangement contains relevant information about shapes, edges, textures and visual patterns.

A traditional Dense neural network would require the image to be flattened into a one-dimensional vector, removing the explicit two-dimensional spatial structure of the image. A **Convolutional Neural Network (CNN)** is better suited to this type of data because convolutional filters operate directly on local regions of the image and preserve spatial relationships while learning relevant visual features.

## 2. Disease classification depends on visual feature learning

Grapevine diseases are identified from visual characteristics of the leaves, including differences in colour, texture, shape and lesion patterns. These relevant features are not explicitly provided as structured variables in the dataset; they must be extracted from the images.

CNNs can learn these visual representations automatically through successive layers. Earlier convolutional layers can detect relatively simple patterns such as edges, colour transitions and textures, while deeper layers can combine them into more complex representations associated with disease symptoms.

This reduces the need to manually define image features in advance and allows the model to learn representations that are directly relevant to the classification task.

## 3. The dataset contains substantial visual variability

The EDA identified considerable variation in leaf orientation, scale, lighting and background. Disease symptoms also vary in appearance and extent, with some visually similar patterns observed across disease classes.

This variability makes it difficult to define a fixed set of handcrafted rules capable of representing all relevant image conditions. A CNN can learn visual representations directly from the available image data and develop features that are more robust to variations in image composition and symptom appearance. However, the reduced effective dataset size and marked class imbalance identified during the extended EDA increase the risk of overfitting and must be considered during model development.

The observed variability therefore supports the use of a learning-based image-classification approach rather than relying on manually specified visual rules.

## 4. CNNs use parameter sharing efficiently

The original images are high-dimensional, containing a large number of pixel values. A fully connected Dense network would require these images to be flattened, leading to a substantial increase in model parameters and preventing the architecture from explicitly exploiting the two-dimensional spatial structure of the data, including the local relationships between neighbouring pixels that can be important for recognising edges, textures and disease symptoms.

CNNs use **local connectivity and parameter sharing**, applying the same learned filters across different regions of an image. This allows the network to detect relevant visual features wherever they appear while using fewer parameters than an equivalent fully connected architecture.

This property makes CNNs particularly appropriate for image data and supports a more efficient learning process for the proposed classification task.

## 5. Why Deep Learning rather than traditional Machine Learning?

Traditional Machine Learning approaches to image classification typically require a separate feature-engineering stage in which relevant image characteristics are manually defined or extracted before classification.

Deep Learning can learn both the relevant representations and the classification function through multiple layers trained directly from image data. In this project, a CNN provides this capability while also exploiting the spatial structure of the images.

For this reason, Deep Learning provides an appropriate framework for the proposed proof of concept, with the CNN serving as the architecture specifically designed for the image-classification task.

## Conclusion

Deep Learning is appropriate for this project because the classification task is based on complex visual information that must be learned directly from images. A CNN is particularly suitable because it can exploit spatial relationships, learn hierarchical visual features and use shared filters efficiently across different image regions.

The EDA also revealed substantial real-world variability in the dataset, reinforcing the need for a model capable of learning visual representations rather than relying on manually defined features or fixed rules.

However, the extended EDA identified substantial exact duplication, reducing the effective modelling dataset from 3,477 original records to **2,407 usable unique images**. The resulting dataset also presents a marked class imbalance, with only **97 Black Rot images**. These limitations increase the risk of overfitting and should be considered when defining the CNN architecture, training strategy and evaluation approach. The resulting model should therefore be evaluated as a **decision-support proof of concept rather than an autonomous diagnostic system**.
